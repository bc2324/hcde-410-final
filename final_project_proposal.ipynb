# HCDE 410 Final Project Proposal

## Motivation & Problem

For my final project, I want to investigate how sports betting companies use YouTube to attract potential customers and promote their services. I am interested in this topic because these companies often combine sports entertainment, memes, and promotional offers with gambling-related content.

My research question is: **What types of content do sports betting companies use on YouTube, and how do they incorporate promotional strategies into sports-related and entertainment content?**

I plan to compare DraftKings and FanDuel to understand how frequently they use promotional offers, sports entertainment, and direct calls to action.

## Data

I plan to study publicly available videos from the official YouTube channels of DraftKings and FanDuel. My primary data source will be YouTube video metadata, accessed through the YouTube Data API (https://developers.google.com/youtube/v3). The API can provide video titles, descriptions, publication dates, and publicly available engagement statistics.

I plan to examine approximately 50–100 videos in total, ideally 25–50 from each company, within a consistent time period. I will categorize videos based on whether they contain promotional offers, sports-related entertainment, direct calls to action, or a combination of these strategies. I will then examine how frequently these strategies appear and how they differ between the two companies. If permitted by the API terms, I may also explore engagement statistics as a secondary part of the analysis.

The data is subject to the YouTube API Services Terms of Service (https://developers.google.com/youtube/terms/api-services-terms-of-service) and Developer Policies (https://developers.google.com/youtube/terms/developer-policies). I will verify the permitted collection, analysis, and storage methods before beginning. Since gambling advertising can influence vulnerable audiences, I also want to consider the ethical implications of these promotional strategies. I will focus on company-published content rather than collecting personal information about individual users.

## Unknowns & Dependencies

My biggest concern is whether I can access enough relevant videos from both companies and whether YouTube's API restrictions will affect my analysis. I will first test data availability and verify the platform's terms. If necessary, I can reduce the sample size or focus on one company.

Another challenge is categorizing videos consistently, since some content may combine entertainment and gambling promotions. To address this, I plan to establish clear definitions for each content category before analyzing the videos.

I also recognize that social media engagement does not necessarily mean someone becomes a paying customer. Because of this, I will avoid making claims about actual customer conversions.

During our in-class discussion, one of my classmates suggested narrowing my original topic from gambling companies in general to sports betting companies specifically because it would make gathering and comparing data easier. I decided to follow this suggestion and adjust my research question accordingly.
